## 1. Prepare verified data

```python
from pathlib import Path
import pandas as pd
import numpy as np

data_path = Path("data/loan_default.csv")  # Update to the approved dataset path.
if not data_path.exists():
    raise FileNotFoundError(f"Dataset not found: {data_path}")
df = pd.read_csv(data_path)
print(df.shape, df.columns.tolist())
display(df.head())
```

## 2. Leakage-aware split and preprocessing

```python
TARGET = None  # Set this only after verifying the supplied dataset documentation.
if TARGET is not None and TARGET in df.columns:
    from sklearn.model_selection import train_test_split
    from sklearn.compose import ColumnTransformer
    from sklearn.pipeline import Pipeline
    from sklearn.impute import SimpleImputer
    from sklearn.preprocessing import OneHotEncoder, StandardScaler
    X = df.drop(columns=[TARGET, "Applicant_ID"], errors="ignore")
    y = df[TARGET]
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42
    )
    num = X_train.select_dtypes(include=np.number).columns.tolist()
    cat = X_train.select_dtypes(exclude=np.number).columns.tolist()
    prep = ColumnTransformer([
        ("num", Pipeline([("impute", SimpleImputer(strategy="median")),
                          ("scale", StandardScaler())]), num),
        ("cat", Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                          ("ohe", OneHotEncoder(handle_unknown="ignore"))]), cat)
    ])
    X_train_t = prep.fit_transform(X_train)
    X_test_t = prep.transform(X_test)
else:
    print("Verify TARGET before training; no target is assumed.")
```
