````python
# Example safety checks for a Text-to-SQL prototype.
# Execute only read-only SELECT queries against an approved SQLite database.
import re
import sqlite3

ALLOWED_PREFIX = re.compile(r"^\s*SELECT\b", re.IGNORECASE)
BLOCKED = re.compile(r"\b(INSERT|UPDATE|DELETE|DROP|ALTER|CREATE|ATTACH|DETACH|PRAGMA|REPLACE)\b", re.IGNORECASE)

def validate_readonly_sql(sql):
    cleaned = sql.strip().rstrip(";")
    if not ALLOWED_PREFIX.match(cleaned):
        raise ValueError("Only SELECT statements are allowed.")
    if BLOCKED.search(cleaned):
        raise ValueError("Potentially unsafe SQL keyword detected.")
    if ";" in cleaned:
        raise ValueError("Multiple statements are not allowed.")
    return cleaned

# Example only; set DB_PATH to an approved database after creating it.
# conn = sqlite3.connect(DB_PATH)
# conn.execute("PRAGMA query_only = ON")
# safe_sql = validate_readonly_sql("SELECT COUNT(*) FROM approved_table")
# result = pd.read_sql_query(safe_sql, conn)
````
