## Safe SQL validation

```python
import re

ALLOWED = re.compile(r"^\\s*SELECT\\b", re.IGNORECASE)
FORBIDDEN = re.compile(
    r"\\b(INSERT|UPDATE|DELETE|DROP|ALTER|CREATE|ATTACH|DETACH|PRAGMA|REPLACE)\\b",
    re.IGNORECASE
)

def validate_readonly_sql(sql):
    statement = sql.strip().rstrip(";")
    if not ALLOWED.match(statement):
        raise ValueError("Only SELECT queries are allowed.")
    if FORBIDDEN.search(statement):
        raise ValueError("Unsafe SQL operation detected.")
    if ";" in statement:
        raise ValueError("Multiple SQL statements are not allowed.")
    return statement

# Before deployment, add schema allow-lists, row limits, timeouts and a
# read-only database connection.
```
